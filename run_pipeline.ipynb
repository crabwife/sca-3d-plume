```python
from pathlib import Path
from time import perf_counter
import os
import subprocess
import sys

import pandas as pd
from IPython.display import display
```

```python
root = Path.cwd()
stages = [
    ('a_import', 'import_data'), ('b_benchmark', 'gaussian_benchmark'),
    ('c_coast', 'coastal_carbon'), ('d_prepare', 'prepare_prairie_grass'),
    ('e_field', 'model_prairie_grass'), ('f_evaluate', 'evaluate'),
]
environment = {**os.environ, 'MPLBACKEND': 'Agg', 'OPENBLAS_NUM_THREADS': '1', 'OMP_NUM_THREADS': '1'}
```

```python
timings = []
for task, name in stages:
    print(f'Running {task}', flush=True)
    source = root / task / 'src' / f'{name}.ipynb'
    command = f'get_ipython().safe_execfile_ipy({source.name!r}, raise_exceptions=True)'
    start = perf_counter()
    subprocess.run([sys.executable, '-m', 'IPython', '--quick', '-c', command],
                   cwd=source.parent, env=environment, check=True)
    timings.append({'task': task, 'seconds': round(perf_counter() - start, 2)})
display(pd.DataFrame(timings))
```
